# Jumia Nigeria Smartphone Market Intelligence
Portfolio case study using the processed listing dataset. **Listings are not sales transactions or market share.**

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from analysis import load_processed, smartphone_scope, market_kpis, brand_summary, segment_summary, channel_summary, engagement_summary
df = load_processed(ROOT / 'data' / 'processed' / 'jumia_phones_clean.csv')
phones = smartphone_scope(df)
market_kpis(df)
```

## Brand concentration
Which brands dominate listing presence?

```python
brands = brand_summary(df)
top10 = brands.head(10).sort_values('listing_share')
fig, ax = plt.subplots(figsize=(9,6))
ax.barh(top10['brand'], top10['listing_share']*100)
ax.set_title('Top 10 Brands by Smartphone Listing Share')
ax.set_xlabel('Listing share (%)')
plt.show()
brands[['brand','listings','listing_share']].head(10)
```

**Finding:** Samsung accounts for about **44.2%** of smartphone-candidate listings; the top five brands together account for about **76.9%**.

## Price-segment structure

```python
segments = segment_summary(df)
segments.assign(listing_share_pct=(segments['listing_share']*100).round(1))
```

**Finding:** Mid-range (₦150k–<₦300k) is the largest project-defined band at about **31.6%**.

## Brand price positioning

```python
major = brands[brands['listings'] >= 20].copy()
major[['brand','listings','median_price_ngn','q1_price_ngn','q3_price_ngn']].head(12)
```

**Finding:** The overall median advertised smartphone price is **₦290,000**; major-brand medians range from value brands such as Itel and Tecno to premium Google and Apple listings.

## Official-store vs third-party presence

```python
channels = channel_summary(df)
channels[['channel','listings','listing_share','median_price_ngn','rating_coverage','average_rating','mean_reviews']]
```

**Finding:** Official-store listings are only **8.3%** of the snapshot, but rating coverage is roughly **77.2%** for official listings versus **29.4%** for third-party listings. This is descriptive, not causal.

## Exploratory engagement associations

```python
engagement_summary(df)
```

Only one-third of in-scope listings have usable engagement data. Correlations are exploratory and should not be interpreted causally.

## Conclusions
1. Listing presence is concentrated around Samsung and four major challengers.
2. The marketplace spans budget through premium positioning, so medians are more useful than means.
3. Third-party sellers dominate listing presence.
4. Data-quality handling materially changes the analysis.
5. Repeated snapshots would unlock price-change and assortment-churn analysis.
